```python
import scipy
import scipy.stats as stats
import pandas as pd
from sklearn.datasets import load_iris
import numpy as np
```

```python
iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df['target'] = iris.target
```

```python
iris_df
```

```python
# Set the hypothesized population mean for Sepal Length
population_mean = 5
sample_mean = iris_df['sepal length (cm)'].mean()
sample_std = iris_df['sepal length (cm)'].std()
n = len(iris_df)

# Perform a one-sample t-test
t_stat, p_value = stats.ttest_1samp(iris_df['sepal length (cm)'], population_mean)  
print("T-statistic:", t_stat)
print("P-value:", p_value)
```
